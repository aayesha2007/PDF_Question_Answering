# Week 13 – PDF Question Answering System

## Introduction
This project develops a PDF Question Answering System using LangChain, ChromaDB, Sentence Transformers, and Ollama. It extracts text from a PDF document, splits the text into smaller chunks, generates embeddings, and stores them in ChromaDB.

The system retrieves relevant information using semantic search and uses a locally running LLM to generate answers based on the retrieved context.

## Objectives
- Extract text from a PDF document.
- Split the extracted text into smaller chunks.
- Generate embeddings using Sentence Transformers.
- Store embeddings in ChromaDB.
- Retrieve the Top-5 relevant chunks for a query.
- Generate answers using Ollama.
- Compare semantic search with keyword-based search.
